Một notebook duy nhất để huấn luyện và tổng hợp kết quả trên Kaggle hoặc Google Colab.
Gồm Static / IRAL / IRAAL cùng hai chiến lược DRMD-FN và DRMD-FN-BHR đã chỉnh cơ chế (V14).
Không cần chạy hai notebook riêng, cũng không phụ thuộc bảng kết quả V13.

### Năm phương pháp × 10 seed = 50 lượt

| Phương pháp | Mô tả |
|---|---|
| **Static-MLP** | MLP cố định trên cửa sổ huấn luyện ban đầu, không cập nhật |
| **DRMD-IRAL** | Phản hồi nội sinh \(Q_t = R_t\), không giới hạn ngân sách |
| **DRMD-IRAAL** | Cập nhật theo ngân sách B = 15 |
| **DRMD-FN** | IRAAL kèm phạt FN thích nghi |
| **DRMD-FN-BHR** | DRMD-FN kèm Balanced Hard Replay |

### Ba chỉnh sửa cho DRMD-FN và DRMD-FN-BHR

1. Phạt FN chỉ áp dụng khi mô hình phân loại trực tiếp. Phần thưởng của hành động từ chối giữ nguyên như DRMD gốc.
2. Bộ điều khiển FN dùng hậu nghiệm Beta; chỉ đổi mức phạt khi khoảng tin cậy đủ hỗ trợ.
3. Với DRMD-FN-BHR, tỷ lệ phát lại mẫu FN giảm theo căn bậc hai của hệ số phạt, để BHR bổ sung cho bộ phạt chứ không nhân đôi tín hiệu lỗi.

Nhãn của chu kỳ \(t\) chỉ được dùng khi fit từ chu kỳ \(t+1\). Tham số cố định: B = 15, chi phí từ chối −0,1, mp = 1, bộ nhớ 5.000 mẫu. Mỗi lượt có 110 chu kỳ kiểm thử và 885.947 mẫu test theo bản LAMDA đã khóa. Báo cáo chỉ hợp lệ khi đủ 50/50 artifact.
